# Agente Analista de Churn Telco
### Powered by LangGraph + Teradata VantageCloud via MCP

Este notebook demuestra un agente conversacional especializado en análisis de churn para empresas de telecomunicaciones.
El agente se conecta en tiempo real a **Teradata VantageCloud** mediante el servidor MCP en Google Cloud Run,
ejecuta SQL in-database y genera insights accionables.

**Dataset:** `DEMO_Telco.Customer_Churn` + `DEMO_Telco.Cdr_Data` — 7,043 clientes

---
```
VS Code (Notebook)  →  LangGraph Agent  →  MCP Server (Cloud Run)  →  Teradata VantageCloud
```

## Setup — Inicializar el Agente

````python
import sys
import os

# Asegura que el directorio actual está en el path
sys.path.insert(0, os.path.dirname(os.path.abspath('__file__')))

from agent import create_analyst

# Inicializa el agente (conecta al MCP server en Cloud Run)
analyst = create_analyst()
````

---
## Acto 1 — Estado Actual del Churn

> *¿Cuántos clientes tenemos y cuál es nuestra tasa de abandono?*

````python
analyst.ask(
    "Explora los datos de Customer_Churn y dame un panorama completo del estado actual del churn: "
    "total de clientes, tasa de abandono, ingreso mensual en riesgo y distribución por tipo de contrato. "
    "Incluye una gráfica de la distribución por contrato."
)
````

---
## Acto 2 — Segmentos de Mayor Riesgo

> *¿Qué tipo de clientes se van más y por qué?*

````python
analyst.ask(
    "Analiza qué segmentos de clientes tienen mayor riesgo de churn. "
    "Revisa por: tipo de servicio de internet, antigüedad del cliente (cohortes de tenure) "
    "y método de pago. ¿Cuáles son los 3 segmentos más críticos?"
)
````

---
## Acto 3 — Señales de Insatisfacción en el Uso

> *¿Qué nos dicen los datos de uso (CDR) sobre los clientes que se van?*

````python
analyst.ask(
    "Usa la tabla Cdr_Data para analizar las señales de insatisfacción. "
    "Específicamente: ¿cómo se relaciona el número de llamadas al servicio al cliente (CustServ Calls) "
    "con la tasa de churn? ¿Qué umbral de llamadas es una señal de alerta roja? "
    "Genera una visualización de esta relación."
)
````

---
## Acto 4 — Perfil del Cliente en Riesgo

> *¿A qué clientes activos debemos contactar hoy?*

````python
analyst.ask(
    "Cruza Customer_Churn con Cdr_Data para identificar los clientes activos (Churn=No) "
    "con mayor probabilidad de abandonar. Busca clientes con perfil de riesgo alto: "
    "contrato mensual, fibra óptica, pocos meses de antigüedad y múltiples llamadas al soporte. "
    "Dame el top 10 con sus datos de contacto y nivel de riesgo."
)
````

---
## Acto 5 — Recomendaciones de Retención

> *¿Qué acciones concretas debemos tomar para reducir el churn?*

````python
analyst.ask(
    "Basándote en todo el análisis anterior, genera un plan de retención con 3 acciones concretas. "
    "Para cada acción indica: segmento objetivo (con número de clientes), "
    "la acción específica a tomar, el ingreso mensual en riesgo que protegemos "
    "y el KPI para medir el éxito. Prioriza por impacto económico."
)
````
